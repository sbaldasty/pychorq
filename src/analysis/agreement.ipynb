# Key agreement

This notebook demonstrates experimentally by repeated simulation, that the BB84, B92, and E91 protocols as implemented produce the same keys for Alice and Bob.

```python
N_TRIALS = 20

alice = Party('alice')
bob = Party('bob')
eve = Party('eve')
source = Party('source')
```

### BB84

Example of two equal keys.

```python
a_key, b_key, _, _ = bb84(alice, bob, eve, 50, 1)
print(a_key)
print(b_key)
```

Repeatedly checking for two equal keys.

```python
mismatches = 0
for _ in range(N_TRIALS):
    a_key, b_key, _, _ = bb84(alice, bob, eve, 50, 1)
    if a_key.val != b_key.val:
        mismatches += 1

print(f'Mismatches: {mismatches} out of {N_TRIALS}')
```
